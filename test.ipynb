```python
import os
import pandas as pd

# On remonte d'un dossier (..) pour atteindre unified_textual_description
base_dir = "../unified_textual_description"

data = []

for root, dirs, files in os.walk(base_dir):
    for file in files:
        if file.endswith(".txt"):
            file_path = os.path.join(root, file)
            
            try:
                with open(file_path, 'r', encoding='utf-8') as f:
                    text_content = f.read().strip()
                    
                    # On stocke le dossier parent et le nom du fichier
                    dossier_parent = os.path.basename(root)
                    source = f"{dossier_parent}/{file}"
                    
                    data.append({
                        "fichier_source": source,
                        "texte_original": text_content
                    })
            except Exception as e:
                print(f"Erreur lors de la lecture de {file_path} : {e}")

df = pd.DataFrame(data)
df["bpmn_genere"] = None

print(f"Nombre total de textes chargés : {len(df)}")
print(df.head())
```

```python
from openai import OpenAI
import pandas as pd
import time

# 1. Initialisation du client
client = OpenAI(
  base_url = "https://integrate.api.nvidia.com/v1",
  api_key = "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"
)

# 2. Définition de la fonction avec métriques
def reformulate_process_text(texte_original):
    system_prompt = """You are an experienced process owner in a leading company.

Considering the context of Business Process Management (BPM) and process modeling:
- Process: Describes a sequence or flow of activities in an organization.

You are given a possibly incomplete process description delimited by '''. Analyze the description, then reformulate it. Your reformulation should be precise, clear and structured.

Follow these instructions:
- Add all possible termination points.
- Ensure the logical order of steps is clear.
- Use standard, professional terminology.

Here are some examples:
<<
Input: 
'''When a manuscript gets the green light for publication, the editors swing into action. If it passes muster, they give a heads-up to the production and design crews.'''

Output: 
The process begins when a manuscript is approved. The editors then review the manuscript. If the manuscript passes the review, the editors notify the production and design teams. If the manuscript does not pass the review, the process ends.
>>"""

    user_prompt = f"Process description: ''' {texte_original} '''"

    # Démarrage du chronomètre
    start_time = time.time()

    try:
        completion = client.chat.completions.create(
            model="meta/llama-3.3-70b-instruct",
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": user_prompt}
            ],
            temperature=0.2,
            top_p=0.7,
            max_tokens=1024,
            stream=False
        )
        
        # Arrêt du chronomètre
        end_time = time.time()
        
        # Extraction des données
        texte_genere = completion.choices[0].message.content.strip()
        finish_reason = completion.choices[0].finish_reason
        
        # Extraction sécurisée des tokens via l'API
        prompt_tokens = completion.usage.prompt_tokens if completion.usage else 0
        completion_tokens = completion.usage.completion_tokens if completion.usage else 0
        total_tokens = completion.usage.total_tokens if completion.usage else 0
        
        return {
            "texte_reformule": texte_genere,
            "latence_secondes": round(end_time - start_time, 2),
            "prompt_tokens": prompt_tokens,
            "completion_tokens": completion_tokens,
            "total_tokens": total_tokens,
            "finish_reason": finish_reason,
            "statut": "Succès"
        }
        
    except Exception as e:
        end_time = time.time()
        return {
            "texte_reformule": None,
            "latence_secondes": round(end_time - start_time, 2),
            "prompt_tokens": None,
            "completion_tokens": None,
            "total_tokens": None,
            "finish_reason": None,
            "statut": f"Erreur: {str(e)}"
        }

# 3. Application sur le DataFrame et collecte des métriques
print("Début de la reformulation et du benchmarking...")

# Liste pour stocker tous les dictionnaires de résultats
resultats = []

for index, row in df.iterrows():
    print(f"Traitement {index + 1}/{len(df)} : {row['fichier_source']}")
    res = reformulate_process_text(row['texte_original'])
    resultats.append(res)
    # Optionnel: time.sleep(0.5) pour éviter de saturer l'API Nvidia si nécessaire

# 4. Intégration des résultats dans le DataFrame
# Convertir la liste de dictionnaires en DataFrame temporaire
df_metrics = pd.DataFrame(resultats)

# Concaténer (joindre) le DataFrame original avec le DataFrame des métriques
df = pd.concat([df, df_metrics], axis=1)

print("Reformulation terminée !")
# Afficher un résumé des performances (ex: latence moyenne)
print(f"\nLatence moyenne : {df['latence_secondes'].mean():.2f} secondes")
print(f"Tokens générés en moyenne : {df['completion_tokens'].mean():.0f}")

# Visualiser les colonnes générées
print(df.head())
```

```python
import pandas as pd
import time
from openai import OpenAI

# 1. Initialisation du client API
client = OpenAI(
  base_url = "https://integrate.api.nvidia.com/v1",
  api_key = "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" 
)

# 2. Création du dataset de test (on isole la première ligne du dataframe complet)
# Vous pouvez changer head(1) par sample(1) si vous voulez une ligne au hasard
triplet_test_df = df.head(1).copy()

# 3. Définition de la fonction d'extraction Zéro-Shot
def extract_triplets_spo(texte_reformule):
    # Sécurité : si la reformulation avait échoué (None ou vide)
    if pd.isna(texte_reformule) or not str(texte_reformule).strip():
        return "Aucun texte à analyser"

    system_prompt = """You are an expert in Process Mining and Natural Language Processing.
Your task is to extract process knowledge from a standardized process description and convert it into structured "Facts".

DEFINITION OF A FACT:
A fact is strictly a Subject-Predicate-Object (SPO) triplet that represents a single action or state in the process.
- Subject: The actor, role, or system performing the action.
- Predicate: The specific action verb (keep it concise).
- Object: The item, document, or entity being acted upon.
Format: (Subject, Predicate, Object)

HANDLING GATEWAYS AND BRANCHES (CRITICAL RULE):
The process description may contain decision points or conditional logic. You must NOT try to model complex logical gateways. 
Instead, you must extract complete, independent execution paths called "Scenarios". 
Whenever a decision point splits the process, you must create separate Scenarios to represent each possible outcome. A scenario traces a sequence of facts from the split point down to a specific termination point.

INSTRUCTIONS:
1. Carefully analyze the provided process description.
2. Identify all actions and the conditions that trigger them.
3. Group the extracted SPO facts into logical Scenarios. 
4. Ensure the chronological order of facts within each scenario is strictly respected.
5. Output ONLY the scenarios and their facts. Do not add conversational filler.

EXPECTED OUTPUT FORMAT:
Scenario 1: [Name of the base path or condition]
1. (Subject, Predicate, Object)
2. (Subject, Predicate, Object)

Scenario 2: [Name of the alternative path]
1. (Subject, Predicate, Object)"""

    user_prompt = f"INPUT PROCESS DESCRIPTION:\n''' {texte_reformule} '''"

    try:
        completion = client.chat.completions.create(
            model="meta/llama-3.3-70b-instruct", # Le "Juge" extracteur
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": user_prompt}
            ],
            temperature=0.1, # Température très basse pour un formatage strict
            top_p=0.7,
            max_tokens=1024,
            stream=False
        )
        return completion.choices[0].message.content.strip()
    except Exception as e:
        return f"Erreur d'extraction : {str(e)}"

# 4. Identification automatique de toutes vos colonnes de reformulation
# On cherche toutes les colonnes qui commencent par "texte_reformule_"
colonnes_reformulees = [col for col in triplet_test_df.columns if col.startswith("texte_reformule_")]

print(f"Début de l'extraction des triplets sur 1 échantillon ({len(colonnes_reformulees)} modèles détectés)...\n")

# 5. Boucle d'extraction
for col_source in colonnes_reformulees:
    print(f"Traitement de la source : {col_source}")
    
    # Création du nouveau nom de colonne selon votre convention
    nom_colonne_extraction = f"{col_source}_extr_llama_3_3_70b"
    
    # On initialise la colonne
    triplet_test_df[nom_colonne_extraction] = None
    
    # Extraction pour la ligne isolée
    texte = triplet_test_df.at[triplet_test_df.index[0], col_source]
    resultat_spo = extract_triplets_spo(texte)
    
    # Enregistrement
    triplet_test_df.at[triplet_test_df.index[0], nom_colonne_extraction] = resultat_spo
    
    # Pause préventive de 2 secondes
    time.sleep(2)

print("\nExtraction terminée avec succès !")

# 6. Affichage brut des résultats générés pour comparer
colonnes_extraites = [col for col in triplet_test_df.columns if "_extr_llama_3_3_70b" in col]

print("\n" + "="*60)
print("COMPARAISON DES EXTRACTIONS PAR MODÈLE SOURCE")
print("="*60)

for col_resultat in colonnes_extraites:
    print(f"\n▶️ RÉSULTAT POUR : {col_resultat}")
    print("-" * 60)
    print(triplet_test_df[col_resultat].iloc[0])
    print("-" * 60)
```
